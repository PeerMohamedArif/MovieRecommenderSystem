# ITEM BASED COLLABORATIVE FILTER WITH ENTIRE DATASET

```python
movietitle_matrix_user_id
```

```python
movie_corr= movietitle_matrix_user_id.corr(method='pearson',min_periods=80)
# pearson : standard correlation coefficient which obtains correlation between all movies
```

```python
movie_corr
```

```python
#create your own dataframe with our own ratings!
givenratings = pd.read_csv("My_Ratings.csv")
givenratings
```

```python
givenratings['Movie Name'][0]
```

```python
similar_movies_list=pd.Series() # CORRELATING OUR MOVIE preference WITH EVERY SINGLE MOVIE IN THE DATASET
for i in range(0,2):
    similar_movie=movie_corr[givenratings['Movie Name'][i]].dropna()
    similar_movie=similar_movie.map(lambda x: x* givenratings ['Ratings'][i] )
    similar_movies_list = pd.concat([similar_movies_list, similar_movie])
```

```python
similar_movies_list.sort_values(inplace = True, ascending = False)
```
